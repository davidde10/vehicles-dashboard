# Análisis exploratorio de datos (EDA)

Exploración básica del conjunto de datos de anuncios de venta de coches (`vehicles_us.csv`) usando **pandas** y **plotly-express**.

```python
import pandas as pd
import plotly.express as px
```

```python
car_data = pd.read_csv('../vehicles_us.csv')
car_data.head()
```

## Información general

```python
car_data.info()
```

```python
car_data.describe()
```

```python
# Valores ausentes por columna
car_data.isna().sum()
```

## Histograma del odómetro

```python
fig = px.histogram(car_data, x='odometer', nbins=50, title='Distribución del odómetro')
fig.show()
```

## Histograma del precio

```python
fig = px.histogram(car_data, x='price', nbins=50, title='Distribución del precio')
fig.show()
```

## Gráfico de dispersión: odómetro vs. precio

```python
fig = px.scatter(car_data, x='odometer', y='price', color='condition', opacity=0.5,
                 title='Odómetro vs. precio')
fig.show()
```

## Precio mediano por tipo de vehículo

```python
price_by_type = car_data.groupby('type')['price'].median().sort_values(ascending=False).reset_index()
fig = px.bar(price_by_type, x='type', y='price', title='Precio mediano por tipo')
fig.show()
```

## Conclusiones

- La mayoría de los coches tienen un kilometraje bajo-medio; hay pocos con valores muy altos.
- El precio tiende a bajar a medida que aumenta el odómetro.
- Los vehículos más grandes (camionetas, pickups) suelen tener precios medianos más altos.
